```python
import random
import statistics
import matplotlib.pyplot as plt
```

### Datos de producción aleatorios para Fabrica 1 y Fabrica 2

```python
# Generar datos de producción aleatorios para la Fabrica 1
fabrica1 = [random.randint(50, 100) for _ in range(100)]

# Generar datos de producción aleatorios para la Fabrica 2
fabrica2 = [random.randint(60, 100) for _ in range(100)]

print("Datos de venta para Fabrica 1 (primeros 100 dias):", fabrica1[:100])
print("Datos de venta para Fabrica 2 (primeros 100 dias):", fabrica2[:100])
```

```python
maximo1 = max(fabrica1)
minimo1 = min(fabrica1)
rango1 = maximo1 - minimo1
media1 = statistics.mean(fabrica1)
mediana1 = statistics.median(fabrica1)
moda1 = statistics.mode(fabrica1)
desviacion_estandar_fabrica1 = statistics.stdev(fabrica1)
print("Estadistica descriptiva de la Fabrica 1 \n", "Media =", media1,
      "\n Mediana =", mediana1, "\n Moda =", moda1, "\n Maximo = ", maximo1,
      "\n Minimo =", minimo1, "\n Rango =", rango1)
print("Desviación estándar =", desviacion_estandar_fabrica1)
```

```python
maximo2 = max(fabrica2)
minimo2 = min(fabrica2)
rango2 = maximo2 - minimo2
media2 = statistics.mean(fabrica2)
mediana2 = statistics.median(fabrica2)
moda2 = statistics.mode(fabrica2)
desviacion_estandar_fabrica2 = statistics.stdev(fabrica2)
print("Estadistica descriptiva de la Fabrica 2 \n", "Media =", media2,
      "\n Mediana =", mediana2, "\n Moda =", moda2, "\n Maximo = ", maximo2,
      "\n Minimo =", minimo2, "\n Rango =", rango2)
print("Desviación estándar =", desviacion_estandar_fabrica2)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize = (8,6))
plt.hist(fabrica1, bins=10, alpha=0.2, color='b', edgecolor = "black")
plt.axvline(media1, color='b', linestyle='dashed', linewidth=2, label=f'Media={media1:.2f}')
plt.axvline(mediana1, color='r', linestyle='dashed', linewidth=2, label=f'Mediana={mediana1:.2f}')
plt.axvline(moda1, color='y', linestyle='dashed', linewidth=2, label=f'Moda={moda1:.2f}')
plt.axvline(media1 - desviacion_estandar_fabrica1, color='black', linestyle='dashed',
             linewidth=2, label=f'Desv. Estándar = {desviacion_estandar_fabrica1:.2f}')
plt.axvline(media1 + desviacion_estandar_fabrica1, color='black', linestyle='dashed',
             linewidth=2)

plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title("Producción diaria de la fabrica 1")
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (8,6))
plt.hist(fabrica2, bins=10, alpha=0.2, color='g', edgecolor = "black")
plt.axvline(media2, color='g', linestyle='dashed', linewidth=2, label=f'Media={media2:.2f}')
plt.axvline(mediana2, color='m', linestyle='dashed', linewidth=2, label=f'Mediana={mediana2:.2f}')
plt.axvline(moda2, color='c', linestyle='dashed', linewidth=2, label=f'Moda={moda2:.2f}')
plt.axvline(media2 - desviacion_estandar_fabrica2, color='black', linestyle='dashed',
             linewidth=2, label=f'Desv. Estándar = {desviacion_estandar_fabrica2:.2f}')
plt.axvline(media2 + desviacion_estandar_fabrica2, color='black', linestyle='dashed',
             linewidth=2)

plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title("Producción diaria de la fabrica 2")
plt.legend()
plt.show()
```
