```python
import pandas as pd

url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-SkillsNetwork/labs/Data%20files/automobileEDA.csv"
df_temp = pd.read_csv(url)
df_temp.to_csv("autos2.csv", index=False)

print("Archivo autos2.csv creado correctamente")
print(df_temp.shape)
```

```python
from statistics import *
import pandas as pd
import matplotlib as plt
import seaborn as sns
from matplotlib import pyplot

autos = pd.read_csv("autos2.csv")
autos
```

Máximo, Mínimo, Moda, Media, Mediana

```python
print("Precio mínimo:", autos.price.min())
print("Precio máximo:", autos.price.max())
print("Media de precios:", autos.price.mean())
print("Mediana de precios:", autos.price.median())
print("Moda de precios:")
print(autos.price.mode())
```

Estadísticos y cuartiles

```python
autos.describe()
```

Desviación estándar

```python
desviacion_estandar_autos = stdev(autos.price)
print("La desviación estándar de los precios de los autos es:", desviacion_estandar_autos)
```

Regla de Sturges

```python
import math

n = len(autos.price.dropna())
k = math.ceil(1 + 3.322 * math.log10(n))
print("Número de bins según la regla de Sturges:", k)

pyplot.hist(autos.price, bins=k)
pyplot.title("Histograma de los precios de los automóviles")
pyplot.xlabel("Precio")
pyplot.ylabel("Frecuencia")
pyplot.show()
```

Boxplot precio vs body-style

```python
sns.boxplot(x="price", y="body-style", data=autos)
pyplot.title("Relación entre precio y tipo de carrocería")
pyplot.xlabel("Precio")
pyplot.ylabel("Tipo de carrocería")
pyplot.show()
```

Boxplot precio vs engine-location

```python
sns.boxplot(x="price", y="engine-location", data=autos)
pyplot.title("Relación entre precio y ubicación del motor")
pyplot.xlabel("Precio")
pyplot.ylabel("Ubicación del motor")
pyplot.show()
```
