Ejemplo2. Tiempo de 100 metros de un grupo de 12 alumnos

```python
#definimos una de las variables llamada tiempos100mts de tipo lista que contiene los tiempos que tardaron los 10 alumnos en correr 100 metros
Tiempos100mts = [7, 6, 9, 5.6, 9.5, 4.5, 8, 7.3, 2, 4.8, 9.6]

#despejamos los valores de la lista
print(Tiempos100mts)
```

```python
# Usamos la funcion mean para determinar el timpo promedio de la lista de valores
print ('El tiempo promedio de lo que corre es: ' + str(mean(Tiempos100mts)))
```

```python
#usamos la funcion median para determinar el tiempo promedio de la lista de valores

#definimos una variable llamada mediana para calcular la madiana de la lista con la funcion median

from statistics import median
mediana = median(Tiempos100mts)
print ('El tiempo promedio de lo que corre es: ' + str(mediana))
```

Funcion miltimode para calcular multiples modas (Por ejemplo, en caso de que sea bimodal)

```python
from statistics import multimode
multimodas = multimode(Tiempos100mts)
print ('La moda de los tiempos es: ' + str(multimodas))
```

Ejemplo 3: Histograma de 2 grupos

```python
#libreria pyton para fabricar valores
import matplotlib.pyplot as plt
```

```python
import random
fabrica1 =[]
for i in range(100):
  numero_aleatorio = round(random.uniform(50,100), 2)
  fabrica1.append(numero_aleatorio)
print(fabrica1)
```

```python
plt.hist(fabrica1)
plt.title("Produccion diaria de la fabrica 1")
```

```python
import random
fabrica1 =[]
for i in range(100):
  numero_aleatorio = round(random.uniform(50,100), 2)
  fabrica1.append(numero_aleatorio)

from statistics import mean, median, mode

media1 = mean (fabrica1)
mediana1 = median (fabrica1)
moda1 = mode (fabrica1)

print("valores de la fabrica 1")
print("la media es=", media1)
print("la mediana es=", mediana1)
print("la moda es=", moda1)
```

```python
import random

fabrica2 =[]
for i in range(100):
  numero_aleatorio = round(random.uniform(40,90), 2)
  fabrica2.append(numero_aleatorio)

print(fabrica2)
```

```python
from statistics import mean, median, mode

media2 = mean (fabrica2)
mediana2 = median (fabrica2)
moda2 = mode (fabrica2)

print("valores de la fabrica 2")
print("la media es=", media2)
print("la mediana es=", mediana2)
print("la moda es=", moda2)
```

```python
import matplotlib.pyplot as plt

plt.hist(fabrica2)
plt.title("Produccion diaria de la fabrica 2")
plt.xlabel("Valores")
plt.ylabel("Frecuencia")
plt.show()
```
